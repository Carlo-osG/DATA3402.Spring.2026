1. Create a `matrix` class with the following properties:
    * It can be initialized in 2 ways:
        1. with arguments `n` and `m`, the size of the matrix. A newly instanciated matrix will contain all zeros.
        2. with a list of lists of values. Note that since we are using lists of lists to implement matrices, it is possible that not all rows have the same number of columns. Test explicitly that the matrix is properly specified.
    * Matrix instances `M` can be indexed with `M[i][j]` and `M[i,j]`.
    * Matrix assignment works in 2 ways:
        1. If `M_1` and `M_2` are `matrix` instances `M_1=M_2` sets the values of `M_1` to those of `M_2`, if they are the same size. Error otherwise.
        2. In example above `M_2` can be a list of lists of correct size.


```python
class matrix:
    def __init__(self, *args):
        
        if len(args) == 2 and all(isinstance(x, int) for x in args):
            n, m = args
            
            if n <= 0 or m <= 0:
                raise ValueError("Matrix dimensions must be positive integers.")
            
            self.data = [[0.0 for _ in range(m)] for _ in range(n)]
        
        
        elif len(args) == 1 and isinstance(args[0], list):
            data = args[0]
            
            if not all(isinstance(row, list) for row in data):
                raise TypeError("Input must be a list of lists.")
            
            row_lengths = [len(row) for row in data]
            if len(set(row_lengths)) != 1:
                raise ValueError("All rows must have the same number of columns.")
            
            self.data = [[float(value) for value in row] for row in data]
        
        
        else:
            raise TypeError("Invalid constructor usage.")
    
    
    def shape(self):
        return (len(self.data), len(self.data[0]))
    
    def __getitem__(self, key):
        if isinstance(key, tuple):
            i, j = key
            return self.data[i][j]
        
        return self.data[key]
    
    
    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.data[i][j] = float(value)
        else:
            self.data[key] = value
    
    
    def assign(self, other):
        
        if isinstance(other, matrix):
            
            if self.shape() != other.shape():
                raise ValueError("Matrix sizes must match for assignment.")
            
            self.data = [row[:] for row in other.data]
        
        
        elif isinstance(other, list):
            
            temp = matrix(other)
            
            if self.shape() != temp.shape():
                raise ValueError("Matrix sizes must match for assignment.")
            
            self.data = [row[:] for row in temp.data]
        
        
        else:
            raise TypeError("Assignment must be matrix or list of lists.")
    
    
    def __repr__(self):
        return f"matrix({self.data})"
```

```python
M1 = matrix(2,3)
print(M1)

M2 = matrix([[1,2,3],[4,5,6]])
print(M2)

print(M2[0][1])
print(M2[0,1])

M2[0,1] = 10
print(M2)

A = matrix([[1,2],[3,4]])
B = matrix([[5,6],[7,8]])

A.assign(B)
print(A)

A.assign([[9,9],[9,9]])
print(A)
```

3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    * `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    * `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    * `eye(n)`: returns the n by n identity matrix.

```python
def constant(n, m, c):
    if n <= 0 or m <= 0:
        raise ValueError("Matrix dimensions must be positive.")
    
    return matrix([[float(c) for _ in range(m)] for _ in range(n)])

def zeros(n, m):
    return constant(n, m, 0.0)

def ones(n, m):
    return constant(n, m, 1.0)

def eye(n):
    if n <= 0:
        raise ValueError("Size must be positive.")
    
    I = zeros(n, n)
    
    for i in range(n):
        I[i, i] = 1.0
    
    return I
```

4. Add the following member functions to your class. Make sure to appropriately test the dimensions of the matrices to make sure the operations are correct.
    * `M.scalarmul(c)`: a matrix that is scalar product $cM$, where every element of $M$ is multiplied by $c$.
    * `M.add(N)`: adds two matrices $M$ and $N$. Don’t forget to test that the sizes of the matrices are compatible for this and all other operations.
    * `M.sub(N)`: subtracts two matrices $M$ and $N$.
    * `M.mat_mult(N)`: returns a matrix that is the matrix product of two matrices $M$ and $N$.
    * `M.element_mult(N)`: returns a matrix that is the element-wise product of two matrices $M$ and $N$.
    * `M.equals(N)`: returns true/false if $M==N$.

```python
    def scalarmul(self, c):
        n, m = self.shape()
        return matrix([[self.data[i][j] * float(c) for j in range(m)] for i in range(n)])


    def add(self, other):
        if not isinstance(other, matrix):
            raise TypeError("Can only add another matrix.")
        
        if self.shape() != other.shape():
            raise ValueError("Matrix dimensions must match for addition.")
        
        n, m = self.shape()
        return matrix([[self.data[i][j] + other.data[i][j] for j in range(m)] for i in range(n)])


    def sub(self, other):
        if not isinstance(other, matrix):
            raise TypeError("Can only subtract another matrix.")
        
        if self.shape() != other.shape():
            raise ValueError("Matrix dimensions must match for subtraction.")
        
        n, m = self.shape()
        return matrix([[self.data[i][j] - other.data[i][j] for j in range(m)] for i in range(n)])


    def mat_mult(self, other):
        if not isinstance(other, matrix):
            raise TypeError("Can only multiply by another matrix.")
        
        n, m = self.shape()
        p, q = other.shape()
        
        if m != p:
            raise ValueError("Incompatible dimensions for matrix multiplication.")
        
        result = matrix(n, q)
        
        for i in range(n):
            for j in range(q):
                total = 0.0
                for k in range(m):
                    total += self.data[i][k] * other.data[k][j]
                result[i, j] = total
        
        return result


    def element_mult(self, other):
        if not isinstance(other, matrix):
            raise TypeError("Can only element-wise multiply another matrix.")
        
        if self.shape() != other.shape():
            raise ValueError("Matrix dimensions must match for element-wise multiplication.")
        
        n, m = self.shape()
        return matrix([[self.data[i][j] * other.data[i][j] for j in range(m)] for i in range(n)])


    def equals(self, other):
        if not isinstance(other, matrix):
            return False
        
        return self.data == other.data
```

```python
    def __add__(self, other):
        return self.add(other)
    
    def __sub__(self, other):
        return self.sub(other)
    
    def __mul__(self, other):
        if isinstance(other, matrix):
            return self.mat_mult(other)
        else:
            return self.scalarmul(other)
    
    def __rmul__(self, other):
        return self.scalarmul(other)
    
    def __eq__(self, other):
        return self.equals(other)
```

6. Demonstrate the basic properties of matrices with your matrix class by creating two 2 by 2 example matrices using your Matrix class and illustrating the following:

$$
(AB)C=A(BC)
$$
$$
A(B+C)=AB+AC
$$
$$
AB\neq BA
$$
$$
AI=A
$$

```python
class matrix:
    def __init__(self, *args):
        if len(args) == 2 and all(isinstance(x, int) for x in args):
            n, m = args
            if n <= 0 or m <= 0:
                raise ValueError("Matrix dimensions must be positive integers.")
            self.data = [[0.0 for _ in range(m)] for _ in range(n)]

        elif len(args) == 1 and isinstance(args[0], list):
            data = args[0]
            if not all(isinstance(row, list) for row in data):
                raise TypeError("Input must be a list of lists.")
            row_lengths = [len(row) for row in data]
            if len(set(row_lengths)) != 1:
                raise ValueError("All rows must have the same number of columns.")
            self.data = [[float(value) for value in row] for row in data]

        else:
            raise TypeError("Invalid constructor usage.")

    def shape(self):
        return (len(self.data), len(self.data[0]))

    def __getitem__(self, key):
        if isinstance(key, tuple):
            i, j = key
            if isinstance(i, slice) or isinstance(j, slice):
                rows = self.data[i]
                return matrix([row[j] for row in rows])
            return self.data[i][j]

        if isinstance(key, slice):
            return matrix(self.data[key])
        return self.data[key]

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self.data[i][j] = float(value)
        else:
            self.data[key] = [float(v) for v in value]

    def assign(self, other):
        if isinstance(other, matrix):
            if self.shape() != other.shape():
                raise ValueError("Matrix sizes must match for assignment.")
            self.data = [row[:] for row in other.data]
        elif isinstance(other, list):
            temp = matrix(other)
            if self.shape() != temp.shape():
                raise ValueError("Matrix sizes must match for assignment.")
            self.data = [row[:] for row in temp.data]
        else:
            raise TypeError("Assignment must be matrix or list of lists.")

    def __repr__(self):
        return f"matrix({self.data})"

    def transpose(self):
        n, m = self.shape()
        return matrix([[self.data[i][j] for i in range(n)] for j in range(m)])

    def row(self, n):
        return matrix([self.data[n]])

    def column(self, n):
        return matrix([[row[n]] for row in self.data])

    def to_list(self):
        return [row[:] for row in self.data]

    @staticmethod
    def constant(n, m, c):
        if n <= 0 or m <= 0:
            raise ValueError("Matrix dimensions must be positive.")
        return matrix([[float(c) for _ in range(m)] for _ in range(n)])

    @staticmethod
    def zeros(n, m):
        return matrix.constant(n, m, 0.0)

    @staticmethod
    def ones(n, m):
        return matrix.constant(n, m, 1.0)

    @staticmethod
    def eye(n):
        if n <= 0:
            raise ValueError("Size must be positive.")
        I = matrix.zeros(n, n)
        for i in range(n):
            I[i, i] = 1.0
        return I

    def scalarmul(self, c):
        n, m = self.shape()
        return matrix([[self.data[i][j] * float(c) for j in range(m)] for i in range(n)])

    def add(self, other):
        if not isinstance(other, matrix):
            raise TypeError("Can only add another matrix.")
        if self.shape() != other.shape():
            raise ValueError("Matrix dimensions must match for addition.")
        n, m = self.shape()
        return matrix([[self.data[i][j] + other.data[i][j] for j in range(m)] for i in range(n)])

    def sub(self, other):
        if not isinstance(other, matrix):
            raise TypeError("Can only subtract another matrix.")
        if self.shape() != other.shape():
            raise ValueError("Matrix dimensions must match for subtraction.")
        n, m = self.shape()
        return matrix([[self.data[i][j] - other.data[i][j] for j in range(m)] for i in range(n)])

    def mat_mult(self, other):
        if not isinstance(other, matrix):
            raise TypeError("Can only multiply by another matrix.")
        n, m = self.shape()
        p, q = other.shape()
        if m != p:
            raise ValueError("Incompatible dimensions for matrix multiplication.")
        result = matrix(n, q)
        for i in range(n):
            for j in range(q):
                total = 0.0
                for k in range(m):
                    total += self.data[i][k] * other.data[k][j]
                result[i, j] = total
        return result

    def element_mult(self, other):
        if not isinstance(other, matrix):
            raise TypeError("Can only element-wise multiply another matrix.")
        if self.shape() != other.shape():
            raise ValueError("Matrix dimensions must match for element-wise multiplication.")
        n, m = self.shape()
        return matrix([[self.data[i][j] * other.data[i][j] for j in range(m)] for i in range(n)])

    def equals(self, other):
        if not isinstance(other, matrix):
            return False
        return self.data == other.data

    def __add__(self, other):
        return self.add(other)

    def __sub__(self, other):
        return self.sub(other)

    def __mul__(self, other):
        if isinstance(other, matrix):
            return self.mat_mult(other)
        else:
            return self.scalarmul(other)

    def __rmul__(self, other):
        return self.scalarmul(other)

    def __eq__(self, other):
        return self.equals(other)
```

```python
A = matrix([[1, 2],
            [3, 4]])

B = matrix([[2, 0],
            [1, 2]])

C = matrix([[0, 1],
            [1, 0]])

I = eye(2)


left1  = (A * B) * C
right1 = A * (B * C)

print("Associativity:")
print("(AB)C =", left1)
print("A(BC) =", right1)
print("Equal?", left1 == right1)
print()


left2  = A * (B + C)
right2 = (A * B) + (A * C)

print("Distributive Property:")
print("A(B+C) =", left2)
print("AB + AC =", right2)
print("Equal?", left2 == right2)
print()


AB = A * B
BA = B * A

print("Non-Commutativity:")
print("AB =", AB)
print("BA =", BA)
print("AB == BA ?", AB == BA)
print()


AI = A * I

print("Identity Property:")
print("AI =", AI)
print("A =", A)
print("Equal?", AI == A)
```
