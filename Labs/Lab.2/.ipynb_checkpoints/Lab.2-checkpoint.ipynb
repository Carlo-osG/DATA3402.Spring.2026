## Responsible Use of Large Language Models (LLMs)

In this lab, **you are allowed and encouraged to use LLMs responsibly** as learning tools.
Think of them as **tutors, reference books, and debugging partners** — not as answer generators.

### Appropriate uses
- Asking for **explanations** of Python concepts (lists, loops, functions, conditionals)
- Getting **hints** or alternative approaches when you are stuck
- Debugging errors *after* you try to reason about them yourself
- Asking an LLM to **explain your own code** back to you

### Not appropriate
- Copy‑pasting complete solutions without understanding them
- Submitting code you cannot explain
- Using an LLM instead of thinking through the problem first

You may be asked to explain your code or reflect briefly on how you used an LLM.

### Commonly used LLMs (examples)

- **ChatGPT** — https://chat.openai.com  
  General‑purpose reasoning, explanations, and debugging. Good for step‑by‑step thinking.

- **Claude** — https://claude.ai  
  Strong at reading longer code and giving structured explanations.

- **Gemini** — https://gemini.google.com  
  Useful for conceptual explanations and comparisons.

- **GitHub Copilot** — https://github.com/features/copilot  
  IDE‑integrated suggestions. Treat suggestions as *ideas*, not answers.

- **Perplexity** — https://www.perplexity.ai  
  Search‑oriented answers with sources; useful for “how does X work?” questions.

No single tool is required or preferred. What matters is **how** you use it.

*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
def check_tic_tac_toe(board):
    n = len(board)
    
    for row in board:
        if row.count(row[0]) == n and row[0] != 0:
            return row[0]
    
    for col in range(n):
        column_vals = [board[row][col] for row in range(n)]
        if column_vals.count(column_vals[0]) == n and column_vals[0] != 0:
            return column_vals[0]
    
    diag1 = [board[i][i] for i in range(n)]
    if diag1.count(diag1[0]) == n and diag1[0] != 0:
        return diag1[0]
    
    diag2 = [board[i][n-1-i] for i in range(n)]
    if diag2.count(diag2[0]) == n and diag2[0] != 0:
        return diag2[0]
    
    for row in board:
        if 0 in row:
            return -1
    
    return 0
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
def place_mark(board, player, row, col, verbose=True):
    size = len(board)
    if not (0 <= row < size and 0 <= col < size):
        if verbose:
            print("Move Is Off The Board.")
        return False
    
    if board[row][col] != 0:
        if verbose:
            print("This Spot Is Taken.")
        return False
    
    board[row][col] = player
    if verbose:
        print(f"Player {player} placed a mark at ({row}, {col}).")
    
    return True
````

````python
board = [[0, 0, 0],
         [0, 0, 0],
         [0, 0, 0]]

place_mark(board, 1, 0, 0)
place_mark(board, 2, 0, 0)
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
row_map = {row_names[i]: i for i in range(len(row_names))}
column_map = {column_names[i]: i for i in range(len(column_names))}

def parse_location_ttt(loc_str, size):
    if not isinstance(loc_str, str) or len(loc_str) != 2:
        print("Bad input. Must be 2 characters like 'A2'.")
        return None
    
    row_char = loc_str[0].upper()
    col_char = loc_str[1]
    
    if row_char not in row_map or row_map[row_char] >= size:
        print("Bad row.")
        return None
    if col_char not in column_map or column_map[col_char] >= size:
        print("Bad column.")
        return None
    
    return row_map[row_char], column_map[col_char]


def nice_place_mark(board, player, loc_str, verbose=True):
    size = len(board)
    loc = parse_location_ttt(loc_str, size)
    
    if loc is None:
        if verbose:
            print("Bad move.")
        return False
    
    row, col = loc
    return place_mark(board, player, row, col, verbose)
````

````python
board = [[0, 0, 0],
         [0, 0, 0],
         [0, 0, 0]]

nice_place_mark(board, 1, "A1")
nice_place_mark(board, 2, "B2")
nice_place_mark(board, 1, "A1")
draw_tic_tac_toe_labeled(board)
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
def tic_tac_toe_game():
    print("**Welcome to Tic-Tac-Toe!**")
    print("---------------------------")
    
    board = [[0 for _ in range(3)] for _ in range(3)]
    
    player = 1
    
    game_over = False
    
    while not game_over:
        draw_tic_tac_toe_labeled(board)
        
        take_ttt_move(board, player)
        
        status = check_tic_tac_toe(board)
        if status != -1:
            game_over = True
            break
        
        player = 2 if player == 1 else 1
    
    draw_tic_tac_toe_labeled(board)
    
    if status == 0:
        print("It's a draw!")
    else:
        print(f"Player {status} wins!")
````

````python
tic_tac_toe_game()
````

*Exercise 10:* Test that your game works for **5×5** tic‑tac‑toe.

````python
def make_ttt_board(n):
    return [[0 for _ in range(n)] for _ in range(n)]
    
def draw_ttt_labeled_dynamic(board):
    n = len(board)
    symbols = {0: " ", 1: "X", 2: "O"}
    
    row_names = list("ABCDEFGHIJKLMNOPQRSTUVWXYZ")
    column_names = list(map(str, range(1, n+1)))
    
    print("  ", end=" ")
    for j in range(n):
        print(column_names[j], end="   ")
    print()
    
    for i in range(n):
        print("  " + " ---" * n)
        
        print(row_names[i], end=" ")
        for j in range(n):
            print(f"| {symbols[board[i][j]]} ", end="")
        print("|")
    
    print("  " + " ---" * n)

def tic_tac_toe_game_dynamic(n=5):
    print(f"Welcome to {n}x{n} Tic-Tac-Toe!")
    print("------------------------")
    
    board = make_ttt_board(n)
    player = 1
    game_over = False
    
    while not game_over:
        draw_ttt_labeled_dynamic(board)
        take_ttt_move(board, player)
        
        status = check_tic_tac_toe(board)
        if status != -1:
            game_over = True
            break
        
        player = 2 if player == 1 else 1
    
    draw_ttt_labeled_dynamic(board)
    if status == 0:
        print("It's a draw!")
    else:
        print(f"Player {status} wins!")
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
def can_win_next(board, player, row, col):
    if board[row][col] != 0:
        return False
    
    board[row][col] = player
    result = check_tic_tac_toe(board)
    board[row][col] = 0  # undo move
    return result == player

import random

def computer_move(board, player):
    size = len(board)
    opponent = 2 if player == 1 else 1
    
    for i in range(size):
        for j in range(size):
            if can_win_next(board, player, i, j):
                place_mark(board, player, i, j)
                print(f"Computer places at {row_names[i]}{j+1} to WIN")
                return
    
    for i in range(size):
        for j in range(size):
            if can_win_next(board, opponent, i, j):
                place_mark(board, player, i, j)
                print(f"Computer places at {row_names[i]}{j+1} to BLOCK")
                return
    
    empty_cells = [(i, j) for i in range(size) for j in range(size) if board[i][j] == 0]
    if empty_cells:
        i, j = random.choice(empty_cells)
        place_mark(board, player, i, j)
        print(f"Computer places at {row_names[i]}{j+1}")

def tic_tac_toe_vs_computer(n=3, human_player=1):
    print(f"Welcome to Tic-Tac-Toe vs Computer ({n}x{n})!")
    board = make_ttt_board(n)
    player = 1
    game_over = False
    
    while not game_over:
        draw_ttt_labeled_dynamic(board)
        
        if player == human_player:
            take_ttt_move(board, player)
        else:
            computer_move(board, player)
        
        status = check_tic_tac_toe(board)
        if status != -1:
            game_over = True
            break
        
        player = 2 if player == 1 else 1
    
    draw_ttt_labeled_dynamic(board)
    if status == 0:
        print("It's a draw!")
    else:
        print(f"Player {status} wins!")
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
def minimax(board, player, depth, max_depth):
    opponent = 2 if player == 1 else 1
    status = check_tic_tac_toe(board)
    
    if status == player:
        return 10 - depth, None
    elif status == opponent:
        return depth - 10, None
    elif status == 0:
        return 0, None
    elif depth >= max_depth:
        return 0, None
    
    best_score = -float('inf')
    best_move = None
    
    size = len(board)
    for i in range(size):
        for j in range(size):
            if board[i][j] == 0:
                board[i][j] = player
                score, _ = minimax(board, opponent, depth + 1, max_depth)
                board[i][j] = 0
                
                score = -score
                if score > best_score:
                    best_score = score
                    best_move = (i, j)
    
    return best_score, best_move

def computer_move_minimax(board, player, max_depth=4):
    score, move = minimax(board, player, depth=0, max_depth=max_depth)
    if move:
        row, col = move
        place_mark(board, player, row, col)
        print(f"Computer places at {row_names[row]}{col+1} using minimax")

def tic_tac_toe_vs_computer_minimax(n=3, human_player=1, max_depth=4):
    print(f"Welcome to Tic-Tac-Toe vs Minimax Computer ({n}x{n})!")
    board = make_ttt_board(n)
    player = 1
    game_over = False
    
    while not game_over:
        draw_ttt_labeled_dynamic(board)
        
        if player == human_player:
            take_ttt_move(board, player)
        else:
            computer_move_minimax(board, player, max_depth)
        
        status = check_tic_tac_toe(board)
        if status != -1:
            game_over = True
            break
        
        player = 2 if player == 1 else 1
    
    draw_ttt_labeled_dynamic(board)
    if status == 0:
        print("It's a draw!")
    else:
        print(f"Player {status} wins!")
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
def computer_vs_computer_game(n=3, max_depth_1=2, max_depth_2=2, verbose=False):
    board = make_ttt_board(n)
    player = 1
    
    while True:
        if player == 1:
            computer_move_minimax(board, player, max_depth_1)
        else:
            computer_move_minimax(board, player, max_depth_2)
        
        status = check_tic_tac_toe(board)
        if status != -1:
            if verbose:
                draw_ttt_labeled_dynamic(board)
                if status == 0:
                    print("Draw")
                else:
                    print(f"Player {status} wins")
            return status
        
        player = 2 if player == 1 else 1


def simulate_games(grid_sizes=[3, 4, 5], num_games=10, max_depth_1=2, max_depth_2=1):
    for n in grid_sizes:
        wins_1 = 0
        wins_2 = 0
        draws = 0
        
        for _ in range(num_games):
            result = computer_vs_computer_game(n, max_depth_1, max_depth_2, verbose=False)
            if result == 1:
                wins_1 += 1
            elif result == 2:
                wins_2 += 1
            else:
                draws += 1
        
        print(f"{n}x{n} board results after {num_games} games:")
        print(f"  Player 1 (smarter) wins: {wins_1}")
        print(f"  Player 2 wins: {wins_2}")
        print(f"  Draws: {draws}")
        win_rate = wins_1 / num_games * 100
        print(f"  Smarter player win rate: {win_rate:.1f}%\n")
````

````python
simulate_games(grid_sizes=[3, 4, 5], num_games=10, max_depth_1=3, max_depth_2=1)
````
